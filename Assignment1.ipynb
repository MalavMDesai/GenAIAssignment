```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
url="https://github.com/MalavMDesai/GenAIAssignment/raw/refs/heads/main/assignment1.csv"
df = pd.read_csv(url)
df
```

```python
corrMatrix = df.corr();
print(corrMatrix)
```

```python
#Interpretation Q1: Which factor most strongly impacts productivity?
target_col = 'Productivity Score'
target_corr = corrMatrix[target_col].drop(target_col).abs()
highest_column = target_corr.idxmax()
highest_value = corrMatrix[target_col][highest_column]
print("Interpretation Q1: Which factor most strongly impacts productivity?")
print(f"Ans 1: \"{highest_column}\" is most strongly impacts productivity with value \"{highest_value}\"")
```

```python
#Splitting features and target(i.e. productivity)
features = df[['Experience (yrs)', 'Training Hours', 'Working Hours', 'Projects']]
target = df['Productivity Score']

#scale the features
scaler = StandardScaler()
features_scaled = scaler.fit_transform(features)
features_scaled
```

```python
feature_train, feature_test, target_train, target_test = train_test_split(
    features_scaled,
    target,
    test_size=0.2,
    random_state=827
)
model = LinearRegression()
model.fit(feature_train, target_train)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
target_pred = model.predict(feature_test)
print("MAE:", mean_absolute_error(target_test, target_pred))
print("MSE:", mean_squared_error(target_test, target_pred))
print("R2 Score:", r2_score(target_test, target_pred))
```
